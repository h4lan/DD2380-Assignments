# **HMM1 · PROBABILITY OF AN OBSERVATION SEQUENCE**


Introduce the same functions used fo HMM0, useful:

```python

def read_matrix(line):
    # The first two numbers give the number of rows and columns.
    tokens = line.split()
    rows = int(tokens[0])
    cols = int(tokens[1])
    values = [float(token) for token in tokens[2:]]

    matrix = []
    for r in range(rows):
        start = r * cols
        end = start + cols
        matrix.append(values[start:end])

    return matrix

def vector_times_matrix(vector, matrix):
    result = []
    for j in range(len(matrix[0])):
        total = 0.0
        for i in range(len(vector)):
            total += vector[i] * matrix[i][j]
        result.append(total)
    return result
```

And create a new one, reallu useful:

```python
def elementwise_product(vector1, vector2):
    result = []
    for i in range(len(vector1)):
        result.append(vector1[i] * vector2[i])
    return result
```

Now define the algorithm:

```python
def forward_probability(A, B, pi, observations):
    
    # Step 1: Initialization, alpha_1(i) = pi[i] * B[i][O_1]
    first_observation = observations[0]
    emission_column = [row[first_observation] for row in B] # Take the column of B for the first observation.
    alpha = elementwise_product(pi, emission_column)

    # Step 2: Recursion
    for observation in observations[1:]:
        # For each current state, add the contributions from all possible previous states.
        predicted = vector_times_matrix(alpha, A)  #alpha_t-1*A

        # Multiply each state value by its emission probability.
        emission_column = [row[observation] for row in B]
        alpha = elementwise_product(predicted, emission_column) #alpha_t= bitwise[(alpha_t-1*A)*B(:,Ot)^T] in vectors 

    # Step 3: Termination. Add the probabilities for all possible final states.
    return sum(alpha)
```

And call it, by reading Kattis matrices, that now include the observations:

### Q3. In the forward recurrence, why may P(Ot = ot | Xt = i, O1:t−1 = o1:t−1) be replaced by P(Ot = ot | Xt = i)? Why must the recurrence still sum over every possible predecessor state?

In an HMM, once I know the current state $X_t=i$, the observation $O_t$ is conditionally independent of the past observations. For example, once I know which coin is used, earlier tosses do not change its probability of heads. I still sum over all previous states $j$ because each one could lead to state $i$. The term $\alpha_{t-1}(j)a_{ji}$ gives the contribution from state $j$, including the probability of the observations seen so far.
