### Q5. Explain why βT (i) = 1. Then explain what γt(i, j) and γt(i) represent and why both are useful when re-estimating an HMM.

At time $T$, there are no future observations, THE remaining sequence is empty and has probability 1, so $\beta_T(i)=1$ for every state $i$.

Given the whole observation sequence, $\gamma_t(i,j)$ is the probability of moving from state $i$ at time $t$ to state $j$ at time $t+1$ AND Baum–Welch uses these expected transition counts to update $A$. Similarly, $\gamma_t(i)$ is the probability of being in state $i$ at time $t$, It weights each observation when updating $B$, and $\gamma_1(i)$ gives the updated initial probability $\pi_i$. This means that there quantities are strictly connected to parameters of the model.

### Q6. Give a concrete convergence rule for your Baum–Welch implementation. Explain how you prevent numerical underflow and how your chosen method changes the computation of the sequence likelihood. Refer to the relevant functions in your code.

In `forward_pass`, we multiply the forward values by `c[t]` at each step so they add up to 1. This helps prevent very small values from being rounded to zero.

In `backward_pass`, we set the last value for each state to `beta[T-1][i] = c[T-1]`. Then we move backwards and multiply the values at each step by `c[t]`: this keeps the backward values consistent with the scaled forward values.

We use the saved factors to calculate the log-likelihood:

$$
\log P(O_{1:T} \mid \lambda) = -\sum_{t=1}^{T} \log c_t.
$$
