**Sensitivity to sequence length:** Our Baum-Welch implementation converges faster on the longer sequence and the parameters are more accurate, compared to the shorter sequence. Having more information (more observations) on the model seems to allow to more accurately evaluate the probabilities to transition from $i$ to $j$ at time $t$ ($\gamma_t(i,j)$) and to be in $i$ at time $t$ ($\gamma_t(i)$), both given the observation sequence and model.

**Likelihood after converging:** The log-likelihood is lower for the longer sequence, which was to be expected: the probability of observing a particular sequence over $T$ time steps $O$ decreases as $T$ increases.
