**Sensitivity to sequence length:** Our Baum-Welch implementation converges faster on the longer sequence, but the parameters are less accurate: the algorithm **fits the example** (observations), but doesn't estimate $A$ or $B$ correctly, compared to the shorter sequence.

For $A$, the starting probabilities $\pi$ influence the expected number of transition from each state: in the initial $\pi_0$, we are more likely to start in state 2 than in the generating model, where $\pi(2) = 0$. This inaccuracy from the starting model seems to influence the estimation in a bad way with longer sequences. 

In particular, the estimation of $B$ seems to suffer a lot more from longer sequences. For example, for the longer one, the probabilities for state 1 are not ordered correctly: observing 0 is more likely than 3, and the likelihood of observing is almost 3 is almost null. The algorithm seems to match the frequencies in the input.

**Likelihood after converging:** We measure the log-likelihood over the sequence length, to get how well the model performs on average, regardless of the length of the observation sequence. This quantity is significantly lower for the longer sequence, which was to be expected, since the model overfitted on the example.
